#### Final data of flood indicators for running the ML models

#### Data includes: Merged data from zenodo (till I-GUIDE), CEEW data, Flow accumulation, Historical flood data, and News data

```python
merged_path = r'/Users/surabhiupadhyay/Documents/Projects/drought_perception_india/data/all_data/RAR_TEAM1_FinalCleanedData_250807.csv'
ceew_path = r'/Users/surabhiupadhyay/Documents/Projects/drought_perception_india/data/all_data/covariates_scaled_district.csv'
flow_acc_path = r'/Users/surabhiupadhyay/Documents/Projects/drought_perception_india/data/all_data/India_FlowAccumulation.csv'
flood_data_path = r'/Users/surabhiupadhyay/Documents/Projects/drought_perception_india/data/all_data/India_HistoricalFloods.csv'
news_path  = r'/Users/surabhiupadhyay/Documents/Projects/drought_perception_india/data/all_data/news_data_2024.csv'
```

```python
import pandas as pd

merged_df = pd.read_csv(merged_path)
ceew_df   = pd.read_csv(ceew_path)
flow_df   = pd.read_csv(flow_acc_path)
flood_df  = pd.read_csv(flood_data_path)
news_df   = pd.read_csv(news_path)
```

```python
for name, df in {
    "merged": merged_df,
    "ceew": ceew_df,
    "flow": flow_df,
    "flood": flood_df,
    "news": news_df
}.items():
    print(f"\n{name.upper()} columns:")
    print(df.columns.tolist())
```

```python
# Make simple merge keys
merged_df["district_key"] = merged_df["district"].astype(str).str.lower().str.strip()
ceew_df["district_key"]   = ceew_df["district_survey"].astype(str).str.lower().str.strip()
news_df["district_key"]   = news_df["District_Clean"].astype(str).str.lower().str.strip()
```

```python
#CEEW: one row per district
ceew_small = (ceew_df
    .groupby("district_key", as_index=False)
    .agg({
        "highersecondaryabove_district": "mean",
        "vulnerability_district": "mean",
        "N": "mean"
    })
)
```

```python
# Flow + Flood: one row per di_code
flow_small  = flow_df[["di_code", "FlowAcc_sum"]].drop_duplicates("di_code")
flood_small = flood_df[["di_code", "GFD_sum"]].drop_duplicates("di_code")
```

```python
# News: keep only Flood, then one row per district
news_df = news_df[news_df["EventType"] == "Flood"].copy()

news_small = (news_df
    .groupby("district_key", as_index=False)
    .agg({
        "SOURCEURLS": "count",
        "Tone_Avg": "mean",
        "Tone_Positive": "mean",
        "Tone_Negative": "mean",
        "Tone_Polarity": "mean",
        "Tone_ActivityRef": "mean",
        "Tone_SelfGroupRef": "mean",
    })
    .rename(columns={"SOURCEURLS": "flood_news_count"})
)
```

```python
# Merge all into survey data
df = merged_df.merge(ceew_small, on="district_key", how="left")
df = df.merge(flow_small, on="di_code", how="left")
df = df.merge(flood_small, on="di_code", how="left")
df = df.merge(news_small, on="district_key", how="left")
```

```python
df.columns.tolist()
```

```python
# columns to remove
cols_to_drop = [
    "district_flood_sentiment",
    "state_flood_sentiment",
    "Unnamed: 0",
    "state.1",
    "state_district_survey.1",
    "district_shapefile", 
    "N",
    "split",
    "n7dy23_recode"
]

# drop safely (won’t error if a column is missing)
df = df.drop(columns=[c for c in cols_to_drop if c in df.columns])
```
